```python
# Missing Value Summary

def missing_summary(df):

    missing = pd.DataFrame({

        "Missing Count":df.isna().sum(),

        "Missing %":round(df.isna().mean()*100,2),

        "Data Type":df.dtypes

    })

    missing = missing[missing["Missing Count"]>0]

    return missing.sort_values(
        by="Missing %",
        ascending=False
    )
```

```python
def classify_missing(df):

    missing = (
        df.isna().mean()*100
    )

    summary = pd.DataFrame({

        "Missing %":missing

    })

    summary["Category"] = pd.cut(

        summary["Missing %"],

        bins=[-1,0,5,20,50,80,100],

        labels=[
            "No Missing",
            "Very Low",
            "Low",
            "Moderate",
            "High",
            "Very High"
        ]

    )

    return summary.sort_values(
        by="Missing %",
        ascending=False
    )
```

Eight player-level variables with more than 70% missing values were removed prior to imputation. Two captain-related variables contained values only for players who served as team captains, meaning the missingness reflected the absence of the role rather than missing observations. Similarly, six goalkeeper-specific punching statistics exhibited more than 70% missing values and highly zero-inflated distributions (median = 0, 75th percentile = 0), indicating that the statistics were only relevant for a small subset of players. Imputing these variables would introduce artificial information and provide little predictive value. Consequently, these variables were excluded before median imputation was applied to the remaining numerical features.

## Logistic Regression

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression

# Preprocessing
preprocessor_lr = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            StandardScaler(),
            numerical_features
        )
    ]
)

# Multinomial Logistic Regression
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor_lr),
        (
            "classifier",
            LogisticRegression(
                max_iter=2000,
                solver="lbfgs",
                multi_class="multinomial",
                random_state=42
            )
        )
    ]
)

# Fit only on the training data
logistic_model.fit(X_train, y_train)

print("Multinomial Logistic Regression fitted successfully.")
```

```python
y_pred_lr = logistic_model.predict(X_test)
y_prob_lr = logistic_model.predict_proba(X_test)

print("Predictions generated successfully.")
print("Probability shape:", y_prob_lr.shape)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Classification metrics
accuracy_lr = accuracy_score(y_test, y_pred_lr)
balanced_accuracy_lr = balanced_accuracy_score(y_test, y_pred_lr)

precision_lr = precision_score(
    y_test, y_pred_lr,
    average="macro",
    zero_division=0
)

recall_lr = recall_score(
    y_test, y_pred_lr,
    average="macro",
    zero_division=0
)

f1_lr = f1_score(
    y_test, y_pred_lr,
    average="macro",
    zero_division=0
)

# Probability-based metrics
log_loss_lr = log_loss(y_test, y_prob_lr)

roc_auc_lr = roc_auc_score(
    y_test,
    y_prob_lr,
    multi_class="ovr",
    average="macro"
)

# Print results
print("LOGISTIC REGRESSION - TEST SET RESULTS")
print("-" * 50)

print(f"Accuracy:           {accuracy_lr:.4f}")
print(f"Balanced Accuracy:  {balanced_accuracy_lr:.4f}")
print(f"Macro Precision:    {precision_lr:.4f}")
print(f"Macro Recall:       {recall_lr:.4f}")
print(f"Macro F1-score:     {f1_lr:.4f}")
print(f"ROC-AUC (Macro):    {roc_auc_lr:.4f}")
print(f"Log Loss:           {log_loss_lr:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_lr,
        target_names=["Home Win (0)", "Draw (1)", "Away Win (2)"],
        digits=4,
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```
