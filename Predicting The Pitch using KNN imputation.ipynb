```python
# KNN imputation function

from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

def knn_impute_dataset(df, n_neighbors=5):
    
    scaler = StandardScaler()
    
    # Standardise the numerical variables
    scaled_data = scaler.fit_transform(df)
    
    # Apply KNN imputation
    imputer = KNNImputer(n_neighbors=n_neighbors)
    imputed_data = imputer.fit_transform(scaled_data)
    
    # Convert back to original scale
    imputed_data = scaler.inverse_transform(imputed_data)
    
    return pd.DataFrame(
        imputed_data,
        columns=df.columns,
        index=df.index
    )
```

```python
# Apply KNN imputation to HOME TEAM

print("Starting KNN imputation for HOME TEAM...")

home_team_num_knn = knn_impute_dataset(
    home_team_num,
    n_neighbors=5
)

print("HOME TEAM KNN imputation completed.")
print("Shape:", home_team_num_knn.shape)
print("Missing values:", home_team_num_knn.isna().sum().sum())
```

```python
# ================================================================
# HOME PLAYER - KNN IMPUTATION
# SINGLE CELL - CLEAN FINAL OUTPUT
# ================================================================

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors


# ---------------------------------------------------------------
# 1. Store original information
# ---------------------------------------------------------------

original_shape = home_player_num.shape
missing_before = home_player_num.isna().sum().sum()


# ---------------------------------------------------------------
# 2. Identify complete and incomplete rows
# ---------------------------------------------------------------

complete_mask = home_player_num.notna().all(axis=1)

complete_data = home_player_num.loc[complete_mask].copy()
incomplete_data = home_player_num.loc[~complete_mask].copy()


# ---------------------------------------------------------------
# 3. Convert complete data to NumPy
# ---------------------------------------------------------------

complete_array = complete_data.to_numpy()

# Column means used temporarily to calculate distances
column_means = complete_data.mean()


# ---------------------------------------------------------------
# 4. Standardise complete observations
# ---------------------------------------------------------------

scaler_player = StandardScaler()

complete_scaled = scaler_player.fit_transform(
    complete_array
)


# ---------------------------------------------------------------
# 5. Fit KNN model
# ---------------------------------------------------------------

knn_player = NearestNeighbors(
    n_neighbors=5,
    metric="euclidean",
    n_jobs=-1
)

knn_player.fit(complete_scaled)


# ---------------------------------------------------------------
# 6. Create output dataset
# ---------------------------------------------------------------

home_player_knn = home_player_num.copy()

incomplete_indices = incomplete_data.index.to_numpy()

batch_size = 500


# ---------------------------------------------------------------
# 7. Perform KNN imputation in batches
# ---------------------------------------------------------------

for start in range(0, len(incomplete_indices), batch_size):

    end = min(
        start + batch_size,
        len(incomplete_indices)
    )

    batch_indices = incomplete_indices[start:end]

    batch = home_player_num.loc[
        batch_indices
    ].copy()


    # Temporarily fill missing values with column means
    batch_filled = batch.fillna(column_means)


    # Standardise
    batch_scaled = scaler_player.transform(
        batch_filled
    )


    # Find 5 nearest complete observations
    distances, neighbour_indices = knn_player.kneighbors(
        batch_scaled
    )


    # Retrieve neighbour values
    neighbour_values = complete_array[
        neighbour_indices
    ]


    # Average the 5 nearest neighbours
    imputed_values = neighbour_values.mean(
        axis=1
    )


    # Convert batch to NumPy
    batch_array = batch.to_numpy()


    # Identify originally missing cells
    missing_mask = np.isnan(batch_array)


    # Replace ONLY missing cells
    batch_array[missing_mask] = (
        imputed_values[missing_mask]
    )


    # Put imputed batch back
    home_player_knn.loc[
        batch_indices,
        home_player_num.columns
    ] = batch_array


# ---------------------------------------------------------------
# 8. Final results
# ---------------------------------------------------------------

missing_after = home_player_knn.isna().sum().sum()


# ---------------------------------------------------------------
# 9. ONE CLEAN OUTPUT
# ---------------------------------------------------------------

print("=" * 70)
print("HOME PLAYER KNN IMPUTATION COMPLETED")
print("=" * 70)

print(f"Original shape        : {original_shape}")
print(f"Complete rows         : {len(complete_data)}")
print(f"Incomplete rows       : {len(incomplete_data)}")
print(f"Missing values before : {missing_before:,}")
print(f"Final shape           : {home_player_knn.shape}")
print(f"Missing values after  : {missing_after:,}")
print(f"KNN neighbours        : 5")

print("=" * 70)
```

```python
# ================================================================
# RESTORE ID + AGGREGATE HOME PLAYER DATA + MERGE
# ================================================================

import pandas as pd

print("=" * 70)
print("PREPARING HOME PLAYER DATA FOR MERGING")
print("=" * 70)

# ------------------------------------------------
# 1. Restore ID to the KNN-imputed Home Player data
# ------------------------------------------------

home_player_knn = home_player_knn.copy()

home_player_knn["ID"] = home_player.loc[
    home_player_knn.index, "ID"
].values

print("ID restored to Home Player KNN data")
print("Home Player KNN shape:", home_player_knn.shape)
print("Unique match IDs:", home_player_knn["ID"].nunique())


# ------------------------------------------------
# 2. Identify numerical columns
# ------------------------------------------------

player_numeric_cols = home_player_knn.select_dtypes(
    include="number"
).columns.tolist()

# Remove ID from variables to aggregate
player_numeric_cols.remove("ID")


# ------------------------------------------------
# 3. Aggregate player statistics by match ID
# ------------------------------------------------
# Each match has multiple players.
# We calculate the mean player statistic for each match.

home_player_agg = (
    home_player_knn
    .groupby("ID")[player_numeric_cols]
    .mean()
    .reset_index()
)

print("Home Player data aggregated")
print("Aggregated shape:", home_player_agg.shape)


# ------------------------------------------------
# 4. Prepare Team datasets
# ------------------------------------------------

home_team_merge = home_team.copy()
away_team_merge = away_team.copy()


# ------------------------------------------------
# 5. Merge Home Team + Away Team
# ------------------------------------------------

merged_knn = pd.merge(
    home_team_merge,
    away_team_merge,
    on="ID",
    suffixes=("_HOME", "_AWAY"),
    how="inner"
)


# ------------------------------------------------
# 6. Merge aggregated Home Player data
# ------------------------------------------------

merged_knn = pd.merge(
    merged_knn,
    home_player_agg,
    on="ID",
    how="left"
)


# ------------------------------------------------
# 7. Final checks
# ------------------------------------------------

print("=" * 70)
print("HOME PLAYER DATA SUCCESSFULLY MERGED")
print("=" * 70)

print("Final shape       :", merged_knn.shape)
print("Unique match IDs  :", merged_knn["ID"].nunique())
print("Duplicate IDs     :", merged_knn["ID"].duplicated().sum())
print("Missing values    :", merged_knn.isna().sum().sum())

print("=" * 70)
```

```python
# ================================================================
# ADD AWAY PLAYER KNN DATA TO MERGED DATASET
# ================================================================

print("=" * 70)
print("ADDING AWAY PLAYER DATA")
print("=" * 70)

# ------------------------------------------------
# 1. Restore ID to Away Player KNN data
# ------------------------------------------------

away_player_knn = away_player_knn.copy()

away_player_knn["ID"] = away_player.loc[
    away_player_knn.index, "ID"
].values

print("Away Player ID restored")
print("Away Player KNN shape:", away_player_knn.shape)
print("Unique match IDs:", away_player_knn["ID"].nunique())


# ------------------------------------------------
# 2. Identify numerical player variables
# ------------------------------------------------

away_player_numeric_cols = away_player_knn.select_dtypes(
    include="number"
).columns.tolist()

away_player_numeric_cols.remove("ID")


# ------------------------------------------------
# 3. Aggregate Away Player statistics by match ID
# ------------------------------------------------

away_player_agg = (
    away_player_knn
    .groupby("ID")[away_player_numeric_cols]
    .mean()
    .reset_index()
)

print("Away Player data aggregated")
print("Aggregated shape:", away_player_agg.shape)


# ------------------------------------------------
# 4. Merge Away Player data
# ------------------------------------------------

merged_knn = pd.merge(
    merged_knn,
    away_player_agg,
    on="ID",
    how="left"
)


# ------------------------------------------------
# 5. Final checks
# ------------------------------------------------

missing_total = merged_knn.isna().sum().sum()

print("=" * 70)
print("AWAY PLAYER DATA SUCCESSFULLY ADDED")
print("=" * 70)

print("Final shape      :", merged_knn.shape)
print("Unique match IDs :", merged_knn["ID"].nunique())
print("Duplicate IDs    :", merged_knn["ID"].duplicated().sum())
print("Missing values   :", missing_total)

print("=" * 70)
```

```python
# ================================================================
# CREATE FINAL FEATURES (X) AND TARGET (y)
# ================================================================

print("=" * 70)
print("CREATING FEATURES AND TARGET")
print("=" * 70)

# ------------------------------------------------
# 1. Create a single target variable
# ------------------------------------------------
# 0 = Home Win
# 1 = Draw
# 2 = Away Win

y_train = y_train.copy()

y_train["TARGET"] = y_train[
    ["HOME_WINS", "DRAW", "AWAY_WINS"]
].idxmax(axis=1)

y_train["TARGET"] = y_train["TARGET"].map({
    "HOME_WINS": 0,
    "DRAW": 1,
    "AWAY_WINS": 2
})


# ------------------------------------------------
# 2. Check target distribution
# ------------------------------------------------

print("Target distribution:")
print(y_train["TARGET"].value_counts().sort_index())


# ------------------------------------------------
# 3. Check that IDs match
# ------------------------------------------------

feature_ids = set(merged_knn["ID"])
target_ids = set(y_train["ID"])

print("\nFeature IDs :", len(feature_ids))
print("Target IDs  :", len(target_ids))
print("IDs match   :", feature_ids == target_ids)


# ------------------------------------------------
# 4. Create X
# ------------------------------------------------
# Remove ID because it is only an identifier
# and should not be used as a predictive feature.

X = merged_knn.drop(columns=["ID"]).copy()


# ------------------------------------------------
# 5. Create y
# ------------------------------------------------
# Match target rows to the same ID order as X.

y = (
    y_train
    .set_index("ID")
    .loc[merged_knn["ID"], "TARGET"]
    .reset_index(drop=True)
)


# ------------------------------------------------
# 6. Final checks
# ------------------------------------------------

print("\n" + "=" * 70)
print("FEATURES AND TARGET CREATED")
print("=" * 70)

print("X shape :", X.shape)
print("y shape :", y.shape)

print("\nTarget counts:")
print(y.value_counts().sort_index())

print("\nMissing values in X :", X.isna().sum().sum())
print("Missing values in y :", y.isna().sum())

print("=" * 70)
```

```python
# ================================================================
# TRAIN-TEST SPLIT
# ================================================================

from sklearn.model_selection import train_test_split

print("=" * 70)
print("CREATING TRAIN AND TEST SETS")
print("=" * 70)

# 80% training, 20% testing
# Stratify keeps the Home Win / Draw / Away Win proportions similar
X_train, X_test, y_train_final, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape :", X_train.shape)
print("X_test shape  :", X_test.shape)
print("y_train shape :", y_train_final.shape)
print("y_test shape  :", y_test.shape)

print("\nTraining target distribution:")
print(y_train_final.value_counts().sort_index())

print("\nTesting target distribution:")
print(y_test.value_counts().sort_index())

print("=" * 70)
```

```python
# ================================================================
# COMMON MODEL EVALUATION FUNCTION
# ================================================================

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    log_loss,
    classification_report,
    confusion_matrix
)

def evaluate_model(model_name, model, X_test, y_test):

    # Predictions
    y_pred = model.predict(X_test)

    # Probabilities
    y_prob = model.predict_proba(X_test)

    # Performance metrics
    accuracy = accuracy_score(y_test, y_pred)

    balanced_acc = balanced_accuracy_score(y_test, y_pred)

    macro_precision = precision_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    macro_recall = recall_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob,
        multi_class="ovr",
        average="macro"
    )

    logloss = log_loss(y_test, y_prob)

    # Print results
    print("=" * 70)
    print(f"{model_name.upper()} - TEST SET RESULTS")
    print("=" * 70)

    print(f"Accuracy:              {accuracy:.4f}")
    print(f"Balanced Accuracy:     {balanced_acc:.4f}")
    print(f"Macro Precision:       {macro_precision:.4f}")
    print(f"Macro Recall:          {macro_recall:.4f}")
    print(f"Macro F1-score:        {macro_f1:.4f}")
    print(f"ROC-AUC (Macro):       {roc_auc:.4f}")
    print(f"Log Loss:              {logloss:.4f}")

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_pred,
            target_names=[
                "Home Win (0)",
                "Draw (1)",
                "Away Win (2)"
            ],
            digits=4,
            zero_division=0
        )
    )

    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    print("=" * 70)

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Balanced Accuracy": balanced_acc,
        "Macro Precision": macro_precision,
        "Macro Recall": macro_recall,
        "Macro F1": macro_f1,
        "ROC-AUC Macro": roc_auc,
        "Log Loss": logloss
    }
```

```python
# ================================================================
# LOGISTIC REGRESSION
# ================================================================

from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

logistic_model.fit(X_train_scaled, y_train_final)

lr_results = evaluate_model(
    "Logistic Regression",
    logistic_model,
    X_test_scaled,
    y_test
)
```

```python
# ================================================================
# RANDOM FOREST
# ================================================================

from sklearn.ensemble import RandomForestClassifier

print("=" * 70)
print("TRAINING RANDOM FOREST")
print("=" * 70)

random_forest = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

random_forest.fit(X_train_scaled, y_train_final)

rf_results = evaluate_model(
    "Random Forest",
    random_forest,
    X_test_scaled,
    y_test
)
```

```python
# ============================================================
# KNN IMPUTATION – AUTOMATIC FINAL MODEL COMPARISON
# ============================================================

import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    log_loss
)
from IPython.display import display


# ------------------------------------------------------------
# Function to calculate all performance metrics
# ------------------------------------------------------------

def calculate_metrics(model_name, model, X_test, y_test):

    # Generate predictions automatically
    y_pred = model.predict(X_test)

    # Generate probabilities automatically
    y_prob = model.predict_proba(X_test)

    return {
        "Imputation": "KNN",
        "Model": model_name,

        "Accuracy": accuracy_score(
            y_test, y_pred
        ),

        "Balanced Accuracy": balanced_accuracy_score(
            y_test, y_pred
        ),

        "Macro Precision": precision_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Macro Recall": recall_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Macro F1": f1_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "ROC-AUC Macro": roc_auc_score(
            y_test,
            y_prob,
            multi_class="ovr",
            average="macro"
        ),

        "Log Loss": log_loss(
            y_test,
            y_prob
        )
    }


# ============================================================
# CALCULATE RESULTS
# ============================================================

knn_results = []


# ------------------------------------------------------------
# Logistic Regression
# ------------------------------------------------------------

knn_results.append(
    calculate_metrics(
        "Logistic Regression",
        logistic_model,
        X_test_scaled,
        y_test
    )
)


# ------------------------------------------------------------
# Decision Tree
# ------------------------------------------------------------

knn_results.append(
    calculate_metrics(
        "Decision Tree",
        decision_tree,
        X_test_scaled,
        y_test
    )
)


# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------

knn_results.append(
    calculate_metrics(
        "Random Forest",
        random_forest,
        X_test_scaled,
        y_test
    )
)


# ------------------------------------------------------------
# XGBoost
# ------------------------------------------------------------

knn_results.append(
    calculate_metrics(
        "XGBoost",
        xgb_model,
        X_test_scaled,
        y_test
    )
)


# ------------------------------------------------------------
# Neural Network / MLP
# ------------------------------------------------------------

knn_results.append(
    calculate_metrics(
        "MLP",
        mlp_model,
        X_test_scaled,
        y_test
    )
)


# ============================================================
# CREATE FINAL COMPARISON TABLE
# ============================================================

knn_results_df = pd.DataFrame(knn_results)


print("=" * 120)
print("KNN IMPUTATION – FINAL MODEL PERFORMANCE COMPARISON")
print("=" * 120)

display(knn_results_df.round(4))
```
